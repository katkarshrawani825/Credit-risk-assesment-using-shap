7. Evaluation Helper Function
   - we build one common function to check any model performance
   - it wiil show accuracy,precision,recall,f!-score and more
   - it will also show a confusion matrux and precision recall curve

```python
# We have to evaluate Logistic Regression and XGBoost models on the basis of their metrics

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report,
    precision_recall_curve
)
import seaborn as sns
import matplotlib.pyplot as plt

def evaluate_model(model_name, model, X_test, y_test, best_threshold):

    # Prediction probabilities
    y_pred_prob = model.predict_proba(X_test)[:, 1]

    # Apply best threshold
    y_pred = (y_pred_prob >= best_threshold).astype(int)

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    print(f"Model: {model_name}")
    print(f"Best Threshold: {best_threshold:.2f}")
    print(f"Accuracy: {accuracy:.2f}")
    print(f"Precision: {precision:.2f}")
    print(f"Recall: {recall:.2f}")
    print(f"F1: {f1:.2f}")
    print()

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)

    print("Confusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, zero_division=0))

    # Plot Confusion Matrix
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"{model_name} Confusion Matrix")
    plt.ylabel("Actual Values")
    plt.xlabel("Predicted Values")
    plt.show()

    # Precision-Recall Curve
    precisions, recalls, thresholds = precision_recall_curve(
        y_test, y_pred_prob
    )

    plt.plot(recalls, precisions)
    plt.title(f"{model_name} Precision-Recall Curve")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.show()

    return y_pred, y_pred_prob, cm
```

#9. Baseline Model - logistic regression#
    -we start with simple model first
    -this becomes our benchmark to beat
    -we train it and check its score on the test data

```python
baseline_model = Pipeline([
    ("Preprocessor", preprocessor_lr),
    ("classifier", LogisticRegression(class_weight="balanced"))

])
baseline_model.fit(X_train,y_train)
```

```python
import numpy as np
from sklearn.metrics import precision_recall_curve

y_prob = baseline_model.predict_proba(X_test)[:, 1]

precision, recall, thresholds = precision_recall_curve(y_test, y_prob)

f1_scores = 2 * (precision * recall) / (precision + recall + 1e-9)

best_index = np.argmax(f1_scores[:-1])
best_threshold = thresholds[best_index]

print("Best Threshold:", best_threshold)
```

```python
evaluate_model("Baseline Logistic Model", baseline_model, X_test, y_test, best_threshold)
```

10. XGBoost hyperparameter tuning
    - Now we move to astronger model,XGboost
    - a model has many setting called hyperparameters
    - we search fro te best combination of these settings
    - we use cross validationwhile searching, so the result is reliable

```python
xg_model=Pipeline([
    ("preprocessor", preprocessor_xg),
    ("classifier", XGBClassifier(n_estimators=300, max_depth=5, learning_rate=0.1, scale_pos_weight=scale_weights, random_state=42))

])
xg_model.fit(X_train, y_train)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform, randint
param_grid = {
    "classifier__n_estimators":randint(150, 600),
     "classifier__max_depth":randint(3, 9),
     "classifier__learning_rate":uniform(0.01, 0.5),
     "classifier__subsample":uniform(0.6,0.4),#rows sampling
     "classifier__colsample_bytree":uniform(0.6, 0.4),#column sampling
    "classifier__min_child_weight":randint(1,10),
    "classifier__gamma":uniform(0,5)
    
}
#lower gamma - more splits -- complex tree -- higher overfitting risk
#higher gamma --  less splits --simpler tree -- low overfitting riskabs
    
    
```

```python
random_search = RandomizedSearchCV(
    xg_model,
    param_grid,
    n_iter=20,
    cv=cv, 
    scoring="average_precision", 
    n_jobs=-1, 
    verbose=1
    
) 
random_search.fit(X_train,y_train)                                
```

12.optimizing the classification threshold
    -the model give a probability, not a direct yes or no
    -we decide a cut-off point to turn it into a decision
    -we check different cut-off points and pick the best one

```python
#get the prob. of class 1 from XGBoost
prob = xg_model.predict_proba(X_test)[:, 1]

#fro each threshold calculate precision and recall 
precision,recalls,threshold = precision_recall_curve(y_test, prob)
threshold
#calculate f1 score for each threshold 
f1 = 2*(precision-recalls)/ (precision + recalls + 1e-9)
f1
# get the threshold that produces the highest f1 score
best = np.argmax(f1[:-1])
best_threshold = threshold[best]
#evaluate the XGBoost model using that threshold 
evaluate_model("XGBoost", xg_model, X_test, y_test, best_threshold)
```

13.Probabaility Calibration
  - sometimes a models probability doesnt mean what it says
  - fro example, 30% risk should truly mean 30% chance
  - we check this with a calibration plot
  - we fix it if the probabilities are off

```python
"""
Methods by which we can fix the poorly calibrated model
1.platt scaling/sigmoid -> adjust the probabilities using a sigmoid curve
2.isotonic regression -> learn a flexible mapping from old prob. ->better prob.(non-linear cali)
3.temperature scaling -> adjust the confidence of neural network prediction using one temperature
"""
```

```python
#1.calibrated XGBoost model
from sklearn.calibration import CalibratedClassifierCV, calibration_curve
best_model = random_search.best_estimator_
calibrated_model = CalibratedClassifierCV(
    best_model,
    method ="sigmoid", 
    cv=5
)
calibrated_model.fit(X_train,y_train)
```

```python
#2.get probabilities
uncalibrated = xg_model.predict_proba(X_test)[:,1]

calibrated = calibrated_model.predict_proba(X_test)[:,1]

#3.create calibration curves
```

15.Analyzing false positive and false negatives
 - the model will not be perfect
 - false positive are safe people wrongly marked risky
 - false negative are risky people erongly marked safe
 - we study these cases to understand where the model struggles
 


```python
#get prediction from the best peforming xgboost model on the test set
y_pred= random_search.predict(X_test)
```

```python
#create a dataframe to easily compare the actual and predicted values
result_df = X_test.copy()
result_df["actual"] = y_test
result_df["predicted"] = y_pred
```

```python
#identify false positive : model predicted 1, but the actualwas 0
false_positive = result_df[(result_df["actual"]==0) & (result_df["predicted"]==1)]

#identify fasle negative : model predicted 0, but the actualwas 1
false_negative = result_df[(result_df["actual"]==1) & (result_df["predicted"]==0)]
```
